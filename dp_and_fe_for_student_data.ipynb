```python
import pandas as pd
import numpy as np
```

```python
df_student = pd.read_csv("student.csv")
df_student
```

```python
df_student['ชื่อ'] = df_student['ชื่อ'].fillna('ไม่ระบุ')
```

## ใช้คำสั่ง fillnaเพื่อแก้ไขชื่อ ที่เป็นNaN เปลี่ยนเป็นคำว่าไม่ระบุแทน


```python
df_student
```

```python
df_student['อายุ'] = df_student['อายุ'].fillna(df_student['อายุ'].median())
df_student
```

## ใช้คำสั่งนี้เพื่อแทนค่าอายุจากค่า Nanเป็นค่ามัธยฐาน

```python
df_student.loc[df_student['อายุ'] > 100, 'อายุ'] = df_student['อายุ'].median()
df_student
```

## ใช้คำสั่งนี้เพื่อแทนอายุที่มากกว่า 100 ปีด้วยค่ามัธยฐาน

```python
df_student['เพศ'] = df_student['เพศ'].fillna('ไม่ระบุ')
df_student['เพศ'] = df_student['เพศ'].replace({
    'ช': 'ชาย', 'ญ': 'หญิง','F':'หญิง','M':'ชาย',
    'Male': 'ชาย', 'Female': 'หญิง'
})
df_student
```

## คำสั่งแรกใช้เพื่อเปลี่ยนข้อมูลเพศที่เป็นค่าNaN เป็นคำว่า ไม่ระบุ และคำสั่งที่สองใช้เพื่อแก้ไขข้อมูลเพศให้เป็นมาตรฐานเดียวกัน

```python
df_student['คณะ'] = df_student['คณะ'].fillna('ไม่ระบุ')
df_student['คณะ'] = df_student['คณะ'].replace({
    'Science': 'วิทยาศาสตร์',
    'วิทย์': 'วิทยาศาสตร์',
    'คณะวิทยาศาสตร์': 'วิทยาศาสตร์',
    'Sci':'วิทยาศาสตร์','Liberal Arts':'ศิลปศาสตร์','ศิลป์':'ศิลปศาสตร์',
    'Arts':'ศิลปศาสตร์','Engineering':'วิศวกรรมศาสตร์','วิศวะ':'วิศวกรรมศาสตร์','วิศวฯ':'วิศวกรรมศาสตร์'
})
df_student
```

## คำสั่งแรกใช้เพื่อเปลี่ยนข้อมูลคณะที่เป็นค่าNaN เป็นคำว่า ไม่ระบุ และคำสั่งที่สองใช้เพื่อแก้ไขข้อมูลคณะต่างๆให้เป็นมาตรฐานเดียวกัน

```python
df_student['GPA'] = df_student['GPA'].fillna(df_student['GPA'].mean())
df_student['GPA'] = df_student['GPA'].clip(lower=0, upper=4)
df_student
```

## คำสั่งแรก แทนค่าเกรดเฉลี่ยที่เป็น NaN ด้วยค่ามัธยฐาน คำสั่งที่สองเป็นการแก้ไขขอบเขตเกรดให้เริ่มที่ 0.00 และไม่เกิน 4.00

```python
df_student['ส่วนสูง_cm'] = df_student['ส่วนสูง_cm'].fillna(df_student['ส่วนสูง_cm'].mean())
df_student
```

## ใช้เพื่อแก้ไขข้อมูลส่วนสูงที่เป็น NaN แทนด้วยค่ามัธยฐาน

```python
df_student['น้ำหนัก_kg'] = df_student['น้ำหนัก_kg'].fillna(df_student['น้ำหนัก_kg'].mean())
df_student
```

## ใช้เพื่อแก้ไขข้อมูลน้ำหนักที่เป็น NaN แทนด้วยค่ามัธยฐาน

```python
df_student['ชั่วโมงเรียน_ต่อสัปดาห์'] = df_student['ชั่วโมงเรียน_ต่อสัปดาห์'].fillna(df_student['ชั่วโมงเรียน_ต่อสัปดาห์'].median())
df_student
```

## ใช้เพื่อแก้ไขข้อมูลชั่วโมงเรียนที่เป็น NaN แทนด้วยค่ามัธยฐาน

```python
df = df_student.drop_duplicates()
df_student
```

## ใช้เพื่อตรวจสอบหาค่าที่ซ่ำกันทุก column และลบทิ้งให้เหลือเพียงแถวเดียว

```python
df['วันที่ลงทะเบียน'] = pd.to_datetime(df['วันที่ลงทะเบียน'], errors='coerce')
df_student
```

## ใช้เพื่อเปลี่ยนชนิดข้อมูลจาก string เป็น datetime เพื่อให้ใช้งานง่ายขึ้น

```python
print("Missing values หลังแก้ไข:")
print(df.isnull().sum())
print("\nข้อมูลหลังแก้ไข:")
print(df.head())
```
